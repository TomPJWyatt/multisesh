```python
import tifffile
import xml.etree.ElementTree as ET

def extract_imagexpress_metadata(file_path):
    # Open the TIFF file and read Tag 270 (ImageDescription)
    with tifffile.TiffFile(file_path) as tif:
        description_xml = tif.pages[0].tags[270].value

    # Parse the embedded XML metadata string
    root = ET.fromstring(description_xml)
    
    extracted_data = {}

    # Standard application properties (Software / Version)
    for prop in root.findall('.//prop'):
        prop_id = prop.get('id')
        if prop_id in ['ApplicationName', 'ApplicationVersion']:
            extracted_data[prop_id] = prop.get('value')

    # ImageXpress hardware identifiers
    target_custom_props = [
        'Instrument Serial Number',
        'ImageXpress Micro Objective',
        'ImageXpress Micro X',
        'ImageXpress Micro Y',
        'ImageXpress Micro Z'
    ]
    for custom_prop in root.findall('.//custom-prop'):
        prop_id = custom_prop.get('id')
        if prop_id in target_custom_props:
            extracted_data[prop_id] = custom_prop.get('value')

    return extracted_data

# Example usage
metadata = extract_imagexpress_metadata(tp)
for key, value in metadata.items():
    print(f"{key}: {value}")
```
